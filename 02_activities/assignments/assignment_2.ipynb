#### Starter Code for Question 1

```python
# Definition for a binary tree node.
class TreeNode(object):
     def __init__(self, val = 0, left = None, right = None):
         self.val = val
         self.left = left
         self.right = right
def is_duplicate(root: TreeNode) -> int:
    if not root:
        return -1
    
    seen = []
    duplicate = -1
    queue = [root]

    while len(queue) > 0:
        node = queue[0]
        queue = queue[1:]

        if node.val in seen:
            if duplicate == -1: 
                duplicate = node.val
        else:
            seen.append(node.val)

        if node.left:
            queue.append(node.left)

        if node.right:
            queue.append(node.right)

    return duplicate 
```
